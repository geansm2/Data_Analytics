## 2. Processamento da Base de Filmes (Metadata)

O arquivo de filmes possui um formato não padrão, exigindo um tratamento customizado para separar o ID, o título e o ano de lançamento.

```python
def load_movie_catalog(filepath):
    # Lendo o arquivo bruto
    # Formato esperado: ID;(Titulo Ano)
    df_movies = pd.read_csv(filepath, sep=',', header=None, names=['raw_data', 'raw_year'])
    
    # Extraindo Movie_Id e Title usando Regex
    # Ex: "1;(Dinosaur Planet" -> ID: 1, Titulo: Dinosaur Planet
    df_movies[['Movie_Id', 'Title']] = df_movies['raw_data'].str.extract(r'(\d+);\((.*)')
    
    # Limpando o Ano (Removendo o parenteses final)
    df_movies['Release_Year'] = df_movies['raw_year'].str.replace(')', '', regex=False)
    
    # Tipagem
    df_movies['Movie_Id'] = pd.to_numeric(df_movies['Movie_Id'])
    
    return df_movies[['Movie_Id', 'Title', 'Release_Year']]

# Exemplo de uso (comentado pois depende do arquivo local)
# movies = load_movie_catalog('movies.csv')
# movies.head()
```
